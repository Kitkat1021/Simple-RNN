## Apply One Hot Encoding

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
# Sentences list
sent=['the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
sent
```

```python
# Define the vocabulary size
voc_size = 10000
```

```python
# One Hot Encode Representation
# List comprehension for onehotencoding
one_hot_repr=[one_hot(words,voc_size)for words in sent]

# See what vectors have been given to the sentences based on the Vocab Size of 10,000
one_hot_repr
```

```python
# Word Embedding Representation
from tensorflow.keras.layers import Embedding

# We need to make all the sentences of same length so we need to add padding to make sure length is same
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
# Set the max length that we require
sent_length = 8

embedded_docs = pad_sequences(
    one_hot_repr,
    padding = 'pre',
    maxlen = sent_length)
```

```python
# We can see that the padding has been added to the front of the sentences
print(embedded_docs)
```
